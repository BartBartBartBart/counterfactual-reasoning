```python
import matplotlib.pyplot as plt
import numpy as np

x_values = np.array([0, 2, 5, 10, 20])
x_positions = np.arange(len(x_values))  # evenly spaced positions
offset = 0.15  # offset for side-by-side plotting

# Create 3 subfigures
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

# AP Exemplar
ap_exemplar_correct_means = [ap_exemplar_probs[str(x)]['correct']['mean'] for x in x_values]
ap_exemplar_correct_stds = [ap_exemplar_probs[str(x)]['correct']['std'] for x in x_values]
ap_exemplar_incorrect_means = [ap_exemplar_probs[str(x)]['incorrect']['mean'] for x in x_values]
ap_exemplar_incorrect_stds = [ap_exemplar_probs[str(x)]['incorrect']['std'] for x in x_values]

axes[0].errorbar(x_positions - offset/2, ap_exemplar_correct_means, yerr=ap_exemplar_correct_stds, 
             fmt='o', color='green', label='Correct', elinewidth=0.5, linewidth=0.5)
axes[0].errorbar(x_positions + offset/2, ap_exemplar_incorrect_means, yerr=ap_exemplar_incorrect_stds, 
             fmt='o', color='red', label='Incorrect', elinewidth=0.5, linewidth=0.5)
axes[0].set_xticks(x_positions)
axes[0].set_xticklabels(x_values)
axes[0].set_ylim(0.35, 1)
axes[0].set_ylabel("Probability")
axes[0].set_xlabel("Number of Permuted Letters")
axes[0].set_title("AP Exemplar")
axes[0].grid(axis='y')
axes[0].legend()

# AP Final Answer
ap_final_correct_means = [ap_final_answer_probs[str(x)]['correct']['mean'] for x in x_values]
ap_final_correct_stds = [ap_final_answer_probs[str(x)]['correct']['std'] for x in x_values] 
ap_final_incorrect_means = [ap_final_answer_probs[str(x)]['incorrect']['mean'] for x in x_values]
ap_final_incorrect_stds = [ap_final_answer_probs[str(x)]['incorrect']['std'] for x in x_values]

axes[1].errorbar(x_positions - offset/2, ap_final_correct_means, yerr=ap_final_correct_stds, 
             fmt='o', color='green', label='Correct', elinewidth=0.5, linewidth=0.5)
axes[1].errorbar(x_positions + offset/2, ap_final_incorrect_means, yerr=ap_final_incorrect_stds, 
             fmt='o', color='red', label='Incorrect', elinewidth=0.5, linewidth=0.5)
axes[1].set_xticks(x_positions)
axes[1].set_xticklabels(x_values)
axes[1].set_ylim(0.35, 1)
axes[1].set_ylabel("Probability")
axes[1].set_xlabel("Number of Permuted Letters")
axes[1].set_title("AP Final Answer")
axes[1].grid(axis='y')
axes[1].legend()

# Baseline Final Answer
baseline_correct_means = [baseline_probs[str(x)]['correct']['mean'] for x in x_values]
baseline_correct_stds = [baseline_probs[str(x)]['correct']['std'] for x in x_values]
baseline_incorrect_means = [baseline_probs[str(x)]['incorrect']['mean'] for x in x_values]
baseline_incorrect_stds = [baseline_probs[str(x)]['incorrect']['std'] for x in x_values]

axes[2].errorbar(x_positions - offset/2, baseline_correct_means, yerr=baseline_correct_stds, 
             fmt='o', color='green', label='Correct', elinewidth=0.5, linewidth=0.5)
axes[2].errorbar(x_positions + offset/2, baseline_incorrect_means, yerr=baseline_incorrect_stds, 
             fmt='o', color='red', label='Incorrect', elinewidth=0.5, linewidth=0.5)
axes[2].set_xticks(x_positions)
axes[2].set_xticklabels(x_values)
axes[2].set_ylim(0.35, 1)
axes[2].set_ylabel("Probability")
axes[2].set_xlabel("Number of Permuted Letters")
axes[2].set_title("Baseline Final Answer")
axes[2].grid(axis='y')
axes[2].legend()

plt.tight_layout()
plt.show()
```
